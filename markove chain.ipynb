```python
import numpy as np
import matplotlib.pyplot as plt

# Transition matrix
T = np.array([
    [0.6, 0.2, 0.2],   # From Up
    [0.3, 0.5, 0.2],   # From Down
    [0.4, 0.3, 0.3]    # From Stable
])

# Initial state (starting in Up)
state = np.array([1, 0, 0])

# Lists to store probabilities
days = [0, 1, 2, 3, 4, 5]
prob_up = [state[0]]
prob_down = [state[1]]
prob_stable = [state[2]]

# Loop for next 5 days
for _ in range(5):
    state = state.dot(T)
    prob_up.append(state[0])
    prob_down.append(state[1])
    prob_stable.append(state[2])

# Plot graph
plt.plot(days, prob_up, label="Up")
plt.plot(days, prob_down, label="Down")
plt.plot(days, prob_stable, label="Stable")

plt.xlabel("Days")
plt.ylabel("Probability")
plt.title("Markov Chain Probabilities Over Time")
plt.legend()
plt.show()
```
